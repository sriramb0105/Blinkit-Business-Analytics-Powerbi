Load Cleaned Data

```python
import pandas as pd
import numpy as np

customers = pd.read_csv("customers_clean.csv")
orders = pd.read_csv("orders_clean.csv")
order_items = pd.read_csv("order_items_clean.csv")
products = pd.read_csv("products_clean.csv")
marketing = pd.read_csv("marketing_clean.csv")
feedback = pd.read_csv("feedback_clean.csv")
```

Merge Datasets

```python
customer_orders = pd.merge(customers,orders,how='left',on='customer_id')
customer_orders
```

```python
customer_orders = customer_orders.merge(order_items,how='left',on='order_id')
customer_orders
```

```python
customer_orders = customer_orders.merge(products,how='left',on='product_id')
customer_orders
```

```python
master_df = customer_orders.merge(feedback,how='left',on='order_id')
master_df
```

Customer Features

```python
customers
```

```python
customers['CLV'] = (customers['total_orders']*customers['avg_order_value'])
```

```python
customers['customer_type'] = np.where(customers['total_orders']>10,'VIP','Regular')
```

```python
customers
```

Order Features

```python
orders
```

```python
orders['Delivery_Delay_Minutes'] = (pd.to_datetime(orders['actual_delivery_time'])-pd.to_datetime(orders['promised_delivery_time'])).dt.total_seconds()/60
```

```python
orders['Late_Delivery_Flag'] = np.where(orders['Delivery_Delay_Minutes']>0,1,0)
```

```python
orders['Order_size'] = pd.cut(orders['order_total'],bins=[0,500,1000,2000,100000],labels=['Small','Medium','Large','Premium'])
```

```python
orders
```

```python
order_items
```

```python
order_items['Revenue'] = (order_items['quantity']*order_items['unit_price'])
```

```python
order_items
```

Product Features

```python
products
```

```python
products['Discount'] = (products['mrp']-products['price'])
```

```python
products['Discount_Percentage'] = (products['Discount']/products['mrp'])*100
```

```python
products['Price_Category'] = pd.cut(products['price'],bins=[0,300,500,700,5000],labels=['Low','Medium','High','Premium'])
```

```python
products
```

Feedback Features

```python
feedback
```

```python
feedback['positive'] = np.where(feedback['rating']>=4,1,0)
```

```python
feedback['negative'] = np.where(feedback['rating']<=2,1,0)
```

```python
feedback['rating_category'] = pd.cut(feedback['rating'],bins=[0,2,3,5],labels=['Poor','Average','Excellent'])
```

```python
feedback
```

Marketing Features

```python
marketing
```

```python
marketing['CTR'] = (marketing['clicks']/marketing['impressions'])*100
```

```python
marketing['Conversion_rate'] = (marketing['conversions']/marketing['clicks'])*100
```

```python
marketing['ROAS'] = (marketing['revenue_generated']/marketing['spend'])
```

```python
marketing['Cost_per_Conversion'] = (marketing['spend']/marketing['conversions'])
```

```python
marketing['Campaign_Profit'] = (marketing['revenue_generated']-marketing['spend'])
```

```python
marketing
```

Create Master Analytical Dataset

```python
master_df = orders.merge(customers,on='customer_id',how='left')
```

```python
master_df = master_df.merge(order_items,on='order_id',how='left')
```

```python
master_df = master_df.merge(products,on='product_id',how='left')
```

```python
master_df = master_df.merge(feedback,on='order_id',how='left')
```

Save Engineered Data

```python
customers.to_csv("customers_features.csv",index=False)
orders.to_csv("orders_features.csv",index=False)
marketing.to_csv("marketing_features.csv",index=False)
products.to_csv("products_features.csv",index=False)
feedback.to_csv("feedback_features.csv",index=False)
order_items.to_csv("order_items_features.csv",index=False)

master_df.to_csv("master_dataset.csv",index=False)
```
