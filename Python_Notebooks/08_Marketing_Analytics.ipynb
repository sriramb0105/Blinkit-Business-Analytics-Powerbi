### Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt

marketing = pd.read_csv("marketing_features.csv")
```

### Marketing KPIs

```python
total_campaigns = marketing["campaign_name"].nunique()

total_spend = marketing["spend"].sum()

total_revenue = marketing["revenue_generated"].sum()

total_profit = marketing["Campaign_Profit"].sum()

average_roas = marketing["ROAS"].mean()

average_ctr = marketing["CTR"].mean()

average_conversion = marketing["Conversion_rate"].mean()
```

```python
print(f"Campaigns : {total_campaigns}")
print(f"Marketing Spend : ₹{total_spend:,.2f}")
print(f"Revenue Generated : ₹{total_revenue:,.2f}")
print(f"Campaign Profit : ₹{total_profit:,.2f}")
print(f"Average ROAS : {average_roas:.2f}")
print(f"Average CTR : {average_ctr:.2f}%")
print(f"Average Conversion Rate : {average_conversion:.2f}%")
```

### Campaign Performance

```python
campaign_revenue = marketing.groupby('campaign_name')['revenue_generated'].sum().sort_values(ascending=False)

campaign_revenue.plot(kind="bar",figsize=(12,5))
plt.title("Revenue by Campaign")
plt.ylabel("Revenue")
plt.show()
```

```python
campaign_profit = marketing.groupby("campaign_name")["Campaign_Profit"].sum().sort_values(ascending=False)

campaign_profit.plot(kind='bar',figsize=(12,5))
plt.title("Profit by Campaign")
plt.ylabel("Profit")
plt.show()
```

### Business Question
Which campaigns generate the highest revenue? Which campaigns generate the highest profit?

### Observation
Referral Program campaign generates the highest revenue and highest profit as well.

### Recommendation
Maintain the spends on top marketing campaigns by profit and optimize changes in spends with the rest.

### Marketing Channel Performance

```python
marketing.groupby("channel")[["spend","revenue_generated","Campaign_Profit"]].sum()
```

```python
channel_spend = (marketing.groupby("channel")["spend"].sum().sort_values(ascending=False))

channel_spend.plot(kind="bar",figsize=(10,5))
plt.title("Spend by Marketing Channel")
plt.show()
```

### Business Question
Which channel does the money being spend most?

### Observation
App Channel is the one where the highest amount is spend at.

```python
channel_revenue = (marketing.groupby("channel")["revenue_generated"].sum().sort_values(ascending=False))

channel_revenue.plot(kind="bar",figsize=(10,5))
plt.title("Revenue by Marketing Channel")
plt.show()
```

### Business Question
Which channel earns the highest revenue?

### Observation
E-mail channel earns the highest revenue, while the App channel contributes second highest revenue though highest amount being spent on the same.

```python
channel_profit = (marketing.groupby("channel")["Campaign_Profit"].sum().sort_values(ascending=False))

channel_profit.plot(kind="bar",figsize=(10,5))
plt.title("Profit by Marketing Channel")
plt.show()
```
